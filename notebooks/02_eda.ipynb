# 02 Exploratory Data Analysis

Analyze pollution distributions, temporal patterns, station comparisons, correlations, and missing data quality.


```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'src').is_dir():
    raise FileNotFoundError('Could not find the project root. Open Jupyter from the project folder.')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import load_all_station_data
from src.preprocessing import clean_air_quality_data
from src.visualization import run_eda_visualizations

raw_dir = PROJECT_ROOT / 'data' / 'raw' / 'Beijing_Multi_Site_Air_Quality'
df = clean_air_quality_data(load_all_station_data(raw_dir))
print(df.shape)
print()
print(df[['station', 'datetime', 'PM2.5', 'PM10', 'SO2']].head())
print()
print(df['PM2.5'].describe())
print()
print(df.isna().mean().sort_values(ascending=False).head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = clean_air_quality_data(load_all_station_data(raw_dir))
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='station', y='PM2.5')
plt.xticks(rotation=45)
plt.title('PM2.5 by Station')
plt.tight_layout()
plt.show()
```

```python
df['hour'] = df['datetime'].dt.hour
hourly = df.groupby('hour')['PM2.5'].mean().reset_index()
hourly.plot(x='hour', y='PM2.5', kind='line', title='Hourly PM2.5 Pattern')
plt.show()
```
