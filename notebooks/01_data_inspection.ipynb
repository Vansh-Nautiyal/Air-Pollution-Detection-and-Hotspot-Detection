# 01 Data Inspection

Inspect the raw Beijing multi-site air quality files before any preprocessing.


```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'src').is_dir():
    raise FileNotFoundError('Could not find the project root. Open Jupyter from the project folder.')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import list_station_files, load_all_station_data
from src.preprocessing import build_datetime_column

raw_dir = PROJECT_ROOT / 'data' / 'raw' / 'Beijing_Multi_Site_Air_Quality'
files = list_station_files(raw_dir)
print(f'Stations found: {len(files)}')
for file_path in files[:5]:
    print(file_path.name)

combined = load_all_station_data(raw_dir)
print(combined.shape)
print(combined.head())
print(combined.columns.tolist())
```

```python
combined = load_all_station_data(raw_dir)
print(combined['station'].value_counts().head())
print()
print(combined.isna().sum().sort_values(ascending=False).head(10))
print()
print(combined['PM2.5'].describe())
```
